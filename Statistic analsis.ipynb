```python
import pandas as pd
df=pd.read_csv(r"E:\Power BI project for ANUDIP\food_delivery_master.csv")
```

```python
from scipy.stats import ttest_ind
peak = df[df['peak_hour'] == 1]['delivery_delay_min'].dropna()
non_peak = df[df['peak_hour'] == 0]['delivery_delay_min'].dropna()

t_stat, p_value = ttest_ind(peak, non_peak)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
from scipy.stats import f_oneway

low = df[df['traffic_level'] == 'Low']['delivery_delay_min'].dropna()
medium = df[df['traffic_level'] == 'Medium']['delivery_delay_min'].dropna()
high = df[df['traffic_level'] == 'High']['delivery_delay_min'].dropna()
f_stat, p_value = f_oneway(low, medium, high)
print("F-statistic:", f_stat)
print("P-value:", p_value)
```
